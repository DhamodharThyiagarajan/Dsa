```python

def bubble_sort(arr):
    n = len(arr)
    for i in range(n):
        swapped = False
        for j in range(0, n - i - 1):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                swapped = True
        if not swapped:
            break
    return arr

# Test the algorithm
print("Bubble Sort:", bubble_sort([64, 34, 25, 12, 22, 11, 90]))
```

Move Zeroes to the End

Problem: Given an integer array, move all 0s to the end while maintaining the relative order of the non-zero elements.
Approach: This uses modified Bubble Sort logic. Instead of bubbling the absolute largest element, you bubble the 0s to the right by swapping them with adjacent non-zero elements.

```python
def move_zeroes(nums):
    n = len(nums)
    # This acts like the optimization flag in Bubble Sort
    last_non_zero_found_at = 0

    # Iterate through the array
    for i in range(n):
        if nums[i] != 0:
            # Swap the non-zero element to the left
            nums[last_non_zero_found_at], nums[i] = nums[i], nums[last_non_zero_found_at]
            last_non_zero_found_at += 1

    return nums

# Test it
print("Move Zeroes:", move_zeroes([0, 1, 0, 3, 12]))
# Expected: [1, 3, 12, 0, 0]
```

# 2. Selection Sort

## Top 5 Interview Questions

**1. Find the Kth Smallest Element**
* **Problem:** Find the Kth smallest element in an unsorted array without fully sorting it.
* **Approach:** Run the outer loop of Selection Sort exactly K times. After K passes, the Kth smallest element will be locked into the `arr[K-1]` position.

**2. Stable Selection Sort**
* **Problem:** Standard Selection Sort is unstable. Modify it so that identical elements maintain their relative initial order.
* **Approach:** Instead of swapping the found minimum element with the first unsorted element, shift all elements from the first unsorted index up to the minimum element's index one position to the right, then place the minimum element in the freed space.

**3. Sort a Linked List using Selection Sort**
* **Problem:** Apply the Selection Sort algorithm to a singly linked list.
* **Approach:** Traverse the linked list with a pointer. For each node, find the minimum node in the remaining list and swap their values.

**4. Maximize Array Sum After K Negations**
* **Problem:** Given an array, negate exactly K elements to maximize the total sum of the array.
* **Approach:** Use a Selection Sort mindset: iteratively find the absolute minimum value in the array K times and multiply it by `-1`.

**5. Sort Half Ascending, Half Descending**
* **Problem:** Sort the first half of an array in ascending order and the second half in descending order.
* **Approach:** Run standard Selection Sort on the first half (finding minimums). Run a modified Selection Sort on the second half (finding maximums).

```python
def selection_sort(arr):
    n = len(arr)
    for i in range(n - 1):
        min_index = i
        for j in range(i + 1, n):
            if arr[j] < arr[min_index]:
                min_index = j
        if min_index != i:
            arr[i], arr[min_index] = arr[min_index], arr[i]
    return arr

# Test the algorithm
print("Selection Sort:", selection_sort([29, 10, 14, 37, 13]))
```

Find the Kth Smallest Element

Problem: Find the Kth smallest element in an unsorted array without fully sorting it.
Approach: Run the outer loop of Selection Sort exactly K times. After K passes, the Kth smallest element will be locked into the arr[K-1] position.

```python
def kth_smallest(arr, k):
    n = len(arr)
    # We only run the loop 'k' times, not 'n' times
    for i in range(k):
        min_index = i
        for j in range(i + 1, n):
            if arr[j] < arr[min_index]:
                min_index = j

        # Swap the found minimum to the front boundary
        arr[i], arr[min_index] = arr[min_index], arr[i]

    # After k passes, the element at k-1 is the Kth smallest
    return arr[k - 1]

# Test it
print("3rd Smallest Element:", kth_smallest([7, 10, 4, 3, 20, 15], 3))
# Expected: 7 (The sorted array would be [3, 4, 7, 10, 15, 20])
```

# 3. Insertion Sort

## Top 5 Interview Questions

**1. Binary Insertion Sort**
* **Problem:** Improve the standard Insertion Sort algorithm to reduce the number of comparisons.
* **Approach:** Instead of linearly searching backward to find the insertion point for the `key`, use Binary Search on the already sorted portion of the array to find the correct index in $O(\log n)$ comparisons.

**2. Sort a Nearly Sorted Array (K-Sorted Array)**
* **Problem:** Sort an array where each element is at most K places away from its target position.
* **Approach:** Insertion Sort is the optimal choice here. Because elements only need to shift a maximum of K times, the time complexity drops from $O(n^2)$ to $O(n \times K)$.

**3. Insertion Sort on a Singly Linked List**
* **Problem:** Implement Insertion Sort for a linked list where you can only traverse forward.
* **Approach:** Create a dummy node representing the sorted list. Iterate through the original list, and for each node, traverse the new sorted list from the head to find its exact insertion point, updating pointers accordingly.

**4. Running Median of a Data Stream**
* **Problem:** Maintain and return the median of a stream of incoming integers dynamically.
* **Approach:** As each new number arrives, insert it into its correct sorted position in an array using Insertion Sort logic so the median can be calculated instantly.

**5. Sort an Array of Objects by Attribute**
* **Problem:** Given an array of Employee dictionaries, sort them by `age` in ascending order.
* **Approach:** Modify the comparison step in the `while` loop to explicitly evaluate `key['age'] < arr[j]['age']`.

```python
def insertion_sort(arr):
    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1
        while j >= 0 and key < arr[j]:
            arr[j + 1] = arr[j]
            j -= 1
        arr[j + 1] = key
    return arr

# Test the algorithm
print("Insertion Sort:", insertion_sort([12, 11, 13, 5, 6]))
```

## 3. Insertion Sort Mechanic: Sort Array by Parity

**Problem:** Given an integer array, move all the even integers at the beginning of the array followed by all the odd integers.

**Why this relates to Insertion Sort:** We can iterate through the array, and whenever we find an even number, we "insert" it at the boundary of our sorted even numbers at the front, dynamically expanding our sorted portion just like Insertion Sort.

```python
def sort_array_by_parity(nums):
    # Pointer to track the end of our "even number" sorted portion
    even_boundary = 0

    for i in range(len(nums)):
        # If the current number is even, swap it into the even boundary
        if nums[i] % 2 == 0:
            nums[even_boundary], nums[i] = nums[i], nums[even_boundary]
            # Expand the boundary
            even_boundary += 1

    return nums

# Test it
print("Sorted by Parity:", sort_array_by_parity([3, 1, 2, 4]))
# Expected: [2, 4, 3, 1] (or any variation where evens are first)
```

# 4. Merge Sort

## Top 5 Interview Questions

**1. Count Inversions in an Array**
* **Problem:** Find how many pairs (i, j) exist such that i < j and `arr[i] > arr[j]`.
* **Approach:** Modify the Merge Sort `merge` phase. Whenever an element from the right half is smaller than an element from the left half, it forms an inversion with all remaining elements in the left half.

**2. Merge Sort on a Linked List**
* **Problem:** Sort a singly linked list in $O(n \log n)$ time and $O(1)$ auxiliary space.
* **Approach:** Use the Fast and Slow pointer technique to find the middle of the linked list, recursively split it, and write a function to merge two sorted linked lists by adjusting pointers.

**3. Merge Two Sorted Arrays In-Place**
* **Problem:** Given two sorted arrays, merge them into one sorted sequence without using extra space.
* **Approach:** Start comparing from the end of the first array and the beginning of the second array. Swap them if they are out of order, then re-sort both arrays.

**4. Reverse Pairs**
* **Problem:** Find the number of important reverse pairs where i < j and `arr[i] > 2 * arr[j]`.
* **Approach:** Use Merge Sort, but add a separate counting step before the standard merge logic to evaluate the `2 *` condition.

**5. Merge K Sorted Lists**
* **Problem:** You are given K sorted linked lists. Merge them into one sorted list.
* **Approach:** The Divide and Conquer approach uses Merge Sort logic: pair up the K lists, merge them into K/2 lists, then K/4 lists, until only one remains.

```python
def merge_sort(arr):
    if len(arr) <= 1:
        return arr
    mid = len(arr) // 2
    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])

    return merge(left, right)

def merge(left, right):
    result = []
    i = j = 0
    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1
    result.extend(left[i:])
    result.extend(right[j:])
    return result

# Test the algorithm
print("Merge Sort:", merge_sort([38, 27, 43, 3, 9, 82, 10]))
```

## 4. Merge Sort Mechanic: Merge Sorted Array (LeetCode 88)
**Problem:** You are given two integer arrays `nums1` and `nums2`, sorted in non-decreasing order. Merge `nums2` into `nums1` as one sorted array. `nums1` has a length of $m + n$, where the first $m$ elements denote the elements that should be merged, and the last $n$ elements are set to `0` and should be ignored.

**Why this relates to Merge Sort:** This is the exact implementation of the "Combine" phase of Merge Sort, but optimized to be done in-place by comparing elements from the back to the front.

```python
def merge_two_arrays(nums1, m, nums2, n):
    # Pointers for nums1, nums2, and the end of the merged array
    p1 = m - 1
    p2 = n - 1
    p = m + n - 1

    # Compare from the back to avoid overwriting elements in nums1
    while p1 >= 0 and p2 >= 0:
        if nums1[p1] > nums2[p2]:
            nums1[p] = nums1[p1]
            p1 -= 1
        else:
            nums1[p] = nums2[p2]
            p2 -= 1
        p -= 1

    # If any elements are left in nums2, copy them over
    while p2 >= 0:
        nums1[p] = nums2[p2]
        p2 -= 1
        p -= 1

    return nums1

# Test it
print("Merged Arrays:", merge_two_arrays([1, 2, 3, 0, 0, 0], 3, [2, 5, 6], 3))
# Expected: [1, 2, 2, 3, 5, 6]
```

# 5. Quick Sort

## Top 5 Interview Questions

**1. Kth Largest Element in an Array (QuickSelect)**
* **Problem:** Find the Kth largest element in an array in $O(n)$ average time.
* **Approach:** Use Quick Sort's partitioning logic. If the pivot lands exactly at index `len(arr) - K`, you have found the element. If the index is smaller, recursively partition the right side; if larger, partition the left side.

**2. Sort an Array of 0s, 1s, and 2s (Dutch National Flag)**
* **Problem:** Sort an array containing only 0, 1, and 2 in one pass with $O(1)$ space.
* **Approach:** This relies on 3-way Quick Sort partitioning. Maintain three pointers (low, mid, high). Evaluate mid: if 0, swap with low; if 1, leave it; if 2, swap with high.

**3. Segregate Positive and Negative Numbers**
* **Problem:** Move all negative numbers to one side of the array and positive numbers to the other.
* **Approach:** Use the Quick Sort partition step with 0 as the conceptual pivot. Iterate through the array and swap any negative number to the left boundary.

**4. Sort Array by Parity (Evens and Odds)**
* **Problem:** Rearrange an array so that all even integers appear before all odd integers.
* **Approach:** Setup two pointers at the start and end of the array. Increment the left if even, decrement the right if odd, and swap them if an odd is found on the left and an even on the right.

**5. Wiggle Sort**
* **Problem:** Sort an array such that `arr[0] <= arr[1] >= arr[2] <= arr[3]...`
* **Approach:** Borrow partition mechanics to swap elements dynamically if they violate the wiggle pattern in a single pass.

```python
# Readable list comprehension version (Great for teaching logic)
def quick_sort(arr):
    if len(arr) <= 1:
        return arr
    pivot = arr[len(arr) // 2]
    left = [x for x in arr if x < pivot]
    middle = [x for x in arr if x == pivot]
    right = [x for x in arr if x > pivot]
    return quick_sort(left) + middle + quick_sort(right)

# Test the algorithm
print("Quick Sort:", quick_sort([10, 7, 8, 9, 1, 5]))
```

## 5. Quick Sort Mechanic: Sort Colors / Dutch National Flag (LeetCode 75)
**Problem:** Given an array with $n$ objects colored red, white, or blue, sort them in-place so that objects of the same color are adjacent, with the colors in the order red, white, and blue. We use the integers `0`, `1`, and `2` to represent the color red, white, and blue, respectively.

**Why this relates to Quick Sort:** This is a famous problem posed by Edsger Dijkstra. It is the perfect application of 3-way Quick Sort partitioning. We use pointers to physically throw `0`s to the left and `2`s to the right, leaving `1`s in the middle.

```python
def sort_colors(nums):
    # low tracks the boundary for 0s, high tracks the boundary for 2s
    low = 0
    mid = 0
    high = len(nums) - 1

    while mid <= high:
        if nums[mid] == 0:
            # Swap to the low boundary and move both pointers forward
            nums[low], nums[mid] = nums[mid], nums[low]
            low += 1
            mid += 1
        elif nums[mid] == 1:
            # 1 is in the correct middle position, just move forward
            mid += 1
        else:
            # Swap to the high boundary and move high pointer backward
            nums[mid], nums[high] = nums[high], nums[mid]
            high -= 1

    return nums

# Test it
print("Sorted Colors:", sort_colors([2, 0, 2, 1, 1, 0]))
# Expected: [0, 0, 1, 1, 2, 2]
```

# Time and Space Complexity Cheat Sheet

| Sorting Algorithm | Best Case Time | Average Case Time | Worst Case Time | Space Complexity |
| :--- | :--- | :--- | :--- | :--- |
| **Bubble Sort** | $O(n)$ | $O(n^2)$ | $O(n^2)$ | $O(1)$ |
| **Selection Sort** | $O(n^2)$ | $O(n^2)$ | $O(n^2)$ | $O(1)$ |
| **Insertion Sort** | $O(n)$ | $O(n^2)$ | $O(n^2)$ | $O(1)$ |
| **Merge Sort** | $O(n \log n)$ | $O(n \log n)$ | $O(n \log n)$ | $O(n)$ |
| **Quick Sort** | $O(n \log n)$ | $O(n \log n)$ | $O(n^2)$ | $O(\log n)$* |

---
**Key Takeaways for Interviews:**
* **$O(1)$ Space:** Bubble, Selection, and Insertion sorts are strictly in-place algorithms requiring no extra memory.
* **Best for Small Data:** Insertion Sort is highly efficient ($O(n)$) on small or nearly sorted datasets.
* **Best Guaranteed Performance:** Merge Sort guarantees $O(n \log n)$ time regardless of the dataset, but at the cost of $O(n)$ extra memory.
* **Best Real-World Performance:** Quick Sort is usually the fastest in practice due to cache locality, but you must mention its $O(n^2)$ worst-case risk during interviews. *(Note: Standard in-place Quick Sort takes $O(\log n)$ space for the recursive call stack. The highly readable list comprehension version provided earlier takes $O(n)$ space).*

# 1. Bubble Sort

## Top 5 Interview Questions

**1. Move Zeroes to the End**
* **Problem:** Given an integer array, move all `0`s to the end while maintaining the relative order of the non-zero elements.
* **Approach:** This uses modified Bubble Sort logic. Instead of bubbling the absolute largest element, you bubble the `0`s to the right by swapping them with adjacent non-zero elements.

**2. Sort an Array in Descending Order**
* **Problem:** Modify standard Bubble Sort to sort elements from highest to lowest.
* **Approach:** Simply flip the comparison operator. If `arr[j] < arr[j + 1]`, swap them. This forces the smallest elements to "bubble" to the end.

**3. Sort an Array of Strings by Length**
* **Problem:** Given an array of strings, sort them so that shorter strings appear first.
* **Approach:** Use the Bubble Sort iteration, but compare the `len(arr[j])` and `len(arr[j+1])` instead of the alphabetical characters.

**4. Check if an Array is Sorted in One Pass**
* **Problem:** Write a function that returns `True` if an array is sorted, using the mechanics of Bubble Sort.
* **Approach:** Utilize the `swapped` flag optimization. Run exactly one pass of the inner loop. If no swaps occur, return `True`. This executes in $O(n)$ time.

**5. Sort by Absolute Difference from a Target Value X**
* **Problem:** Given an array and a value X, sort the array based on the absolute difference between the elements and X.
* **Approach:** During the adjacent element comparison step, evaluate `abs(arr[j] - X) > abs(arr[j+1] - X)`. If true, swap the elements.
